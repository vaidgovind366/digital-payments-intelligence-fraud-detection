```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
transactions = pd.read_csv("transactions_100k.csv")
customers = pd.read_csv("customers.csv")

print("Transactions Shape:", transactions.shape)
print("Customers Shape:", customers.shape)
```

```python
print("TRANSACTIONS")
display(transactions.head())

print("\nCUSTOMERS")
display(customers.head())
```

```python
transactions.info()
```

```python
missing_tx = transactions.isnull().sum()

missing_tx[missing_tx > 0]
```

```python
print("Duplicate Rows:", transactions.duplicated().sum())
print(
    "Duplicate Transaction IDs:",
    transactions["Transaction_ID"].duplicated().sum()
)
print(
    "Duplicate Customer IDs:",
    customers["Customer_ID"].duplicated().sum()
)
```

```python
transactions["Transaction_Timestamp"] = pd.to_datetime(
    transactions["Transaction_Timestamp"]
)

transactions["Transaction_Date"] = pd.to_datetime(
    transactions["Transaction_Date"]
)

transactions["Previous_Transaction_Time"] = pd.to_datetime(
    transactions["Previous_Transaction_Time"],
    errors="coerce"
)

customers["Customer_Since"] = pd.to_datetime(
    customers["Customer_Since"]
)
```

```python
transactions["Year"] = transactions["Transaction_Date"].dt.year
transactions["Month"] = transactions["Transaction_Date"].dt.month
transactions["Month_Name"] = transactions["Transaction_Date"].dt.month_name()
transactions["Day"] = transactions["Transaction_Date"].dt.day
transactions["Hour"] = transactions["Transaction_Timestamp"].dt.hour
transactions["Day_Name"] = transactions["Transaction_Date"].dt.day_name()
```

```python
transactions["Amount"].describe()
```

```python
fraud_summary = transactions.groupby("Fraud_Flag").agg(
    Transactions=("Transaction_ID", "count"),
    Total_Amount=("Amount", "sum"),
    Average_Amount=("Amount", "mean"),
    Average_Risk_Score=("Risk_Score", "mean")
).reset_index()

fraud_summary
```

```python
fraud_reason_analysis = (
    transactions[transactions["Fraud_Flag"] == 1]
    .groupby("Fraud_Reason")
    .agg(
        Fraud_Transactions=("Transaction_ID", "count"),
        Fraud_Amount=("Amount", "sum"),
        Avg_Risk_Score=("Risk_Score", "mean")
    )
    .reset_index()
    .sort_values("Fraud_Transactions", ascending=False)
)

fraud_reason_analysis
```

```python
customer_analysis = (
    transactions.groupby("Customer_ID")
    .agg(
        Total_Transactions=("Transaction_ID", "count"),
        Total_Spend=("Amount", "sum"),
        Avg_Transaction=("Amount", "mean"),
        Successful_Transactions=("Transaction_Status", 
                                   lambda x: (x == "Successful").sum()),
        Failed_Transactions=("Transaction_Status",
                             lambda x: (x == "Failed").sum()),
        Fraud_Transactions=("Fraud_Flag", "sum"),
        Max_Risk_Score=("Risk_Score", "max")
    )
    .reset_index()
)

customer_analysis.head()
```

```python
customer_360 = transactions.merge(
    customers,
    on="Customer_ID",
    how="left",
    validate="many_to_one"
)

print(customer_360.shape)
customer_360.head()
```

```python
fraud_counts = (
    transactions["Fraud_Flag"]
    .value_counts()
    .rename(index={0: "Non-Fraud", 1: "Fraud"})
)

print(fraud_counts)
```

```python
fraud_rate = transactions["Fraud_Flag"].mean() * 100

print(f"Overall Fraud Rate: {fraud_rate:.2f}%")
```

```python
fraud_reason = (
    transactions[transactions["Fraud_Flag"] == 1]
    .groupby("Fraud_Reason")
    .agg(
        Fraud_Count=("Transaction_ID", "count"),
        Fraud_Amount=("Amount", "sum"),
        Avg_Risk_Score=("Risk_Score", "mean")
    )
    .sort_values("Fraud_Count", ascending=False)
    .reset_index()
)

fraud_reason
```

```python
behavior_columns = {
    "New Device": "New_Device_Flag",
    "Location Mismatch": "Location_Mismatch_Flag",
    "Unusual Hour": "Unusual_Hour_Flag",
    "IP Mismatch": "IP_Mismatch_Flag",
    "Failed Burst": "Failed_Burst_Flag",
    "High Velocity": "Velocity_10Min_Count"
}

behavior_summary = []

for name, column in behavior_columns.items():

    if column == "Velocity_10Min_Count":
        flagged = transactions[column] >= 3
    else:
        flagged = transactions[column] == 1

    total_flagged = flagged.sum()
    fraud_flagged = transactions.loc[flagged, "Fraud_Flag"].sum()

    fraud_rate = (
        fraud_flagged / total_flagged * 100
        if total_flagged > 0 else 0
    )

    behavior_summary.append({
        "Behavior": name,
        "Flagged_Transactions": total_flagged,
        "Fraud_Transactions": fraud_flagged,
        "Fraud_Rate_%": round(fraud_rate, 2)
    })

behavior_summary = pd.DataFrame(behavior_summary)

behavior_summary
```

```python
amount_analysis = transactions[
    ["Transaction_ID",
     "Customer_ID",
     "Amount",
     "Customer_Avg_Amount",
     "Amount_Deviation_Pct",
     "Fraud_Flag"]
].copy()

amount_analysis["Amount_vs_Baseline"] = (
    amount_analysis["Amount"] /
    amount_analysis["Customer_Avg_Amount"]
)

amount_analysis.sort_values(
    "Amount_Deviation_Pct",
    ascending=False
).head(20)
```

```python
high_amount = transactions[
    transactions["Fraud_Reason"] == "High Amount Anomaly"
]

print("High Amount Fraud Transactions:", len(high_amount))
print(
    "Average Fraud Amount:",
    round(high_amount["Amount"].mean(), 2)
)
print(
    "Average Deviation %:",
    round(high_amount["Amount_Deviation_Pct"].mean(), 2)
)
```

```python
customer_risk = (
    transactions.groupby("Customer_ID")
    .agg(
        Total_Transactions=("Transaction_ID", "count"),
        Total_Spend=("Amount", "sum"),
        Avg_Transaction=("Amount", "mean"),
        Fraud_Transactions=("Fraud_Flag", "sum"),
        Max_Risk_Score=("Risk_Score", "max"),
        Avg_Risk_Score=("Risk_Score", "mean")
    )
    .reset_index()
)

customer_risk.head()
```

```python
def classify_customer_risk(score):
    if score >= 61:
        return "High"
    elif score >= 31:
        return "Medium"
    else:
        return "Low"


customer_risk["Customer_Risk_Level"] = (
    customer_risk["Max_Risk_Score"]
    .apply(classify_customer_risk)
)

customer_risk["Customer_Risk_Level"].value_counts()
```

```python
top_high_risk = (
    customer_risk[
        customer_risk["Customer_Risk_Level"] == "High"
    ]
    .sort_values(
        ["Max_Risk_Score", "Fraud_Transactions", "Total_Spend"],
        ascending=False
    )
    .head(10)
)

top_high_risk
```

```python
monthly_fraud = (
    transactions.groupby(["Year", "Month"])
    .agg(
        Total_Transactions=("Transaction_ID", "count"),
        Fraud_Transactions=("Fraud_Flag", "sum"),
        Total_Amount=("Amount", "sum"),
        Fraud_Amount=(
            "Amount",
            lambda x: x[
                transactions.loc[x.index, "Fraud_Flag"] == 1
            ].sum()
        )
    )
    .reset_index()
)

monthly_fraud["Fraud_Rate_%"] = (
    monthly_fraud["Fraud_Transactions"] /
    monthly_fraud["Total_Transactions"] * 100
)

monthly_fraud.head(12)
```

```python
investigation_queue = (
    transactions[
        transactions["Fraud_Flag"] == 1
    ][
        [
            "Transaction_ID",
            "Customer_ID",
            "Amount",
            "Payment_Method",
            "Location",
            "Risk_Score",
            "Risk_Level",
            "Fraud_Reason",
            "Transaction_Date",
            "Transaction_Time"
        ]
    ]
    .sort_values(
        ["Risk_Score", "Amount"],
        ascending=False
    )
)

investigation_queue.head(20)
```

```python
fraud_reason.to_csv(
    "python_fraud_reason_analysis.csv",
    index=False
)

behavior_summary.to_csv(
    "python_behavior_analysis.csv",
    index=False
)

customer_risk.to_csv(
    "python_customer_risk_analysis.csv",
    index=False
)

monthly_fraud.to_csv(
    "python_monthly_fraud_analysis.csv",
    index=False
)

investigation_queue.to_csv(
    "python_investigation_queue.csv",
    index=False
)

print("All Python analysis files exported successfully.")
```
